```python
import numpy as np
arr1 = np.array([9, 4, 16])
arr2= np.sqrt(arr1)
arr3 = arr2.astype(int)
print(arr3)
```

# axis=1 in column  and the axis=0 rows 

```python
arr1 = np.array([[9, 4, 16],[25,49,64]])
arr2= np.sqrt(arr1)
arr3 = np.concatenate((arr1, arr2),axis=0).astype(int)
print(arr3)
```

```python
arr1 = np.array([[9, 4, 16],[25,49,64]])
arr2= np.sqrt(arr1)
print(np.vstack((arr1, arr2)).astype(int))
```

```python
arr1 = np.array([[9, 4, 16],[25,49,64]])
arr2= np.sqrt(arr1)
arr3 = np.concatenate((arr1, arr2),axis=1).astype(int)
print(arr3)
```

```python
arr1 = np.array([[9, 4, 16],[25,49,64]])
arr2= np.sqrt(arr1)
print(np.hstack((arr1, arr2)).astype(int))
```

```python
import numpy
a= np.array([[20,40,30,40,10,20],[40,30,50,10,20,40]])
b= np.array([[40,30,60,10,20,40],[90,40,30,40,10,20]])
print(a,"\n",b)

c= np.cumprod([a,b], axis=1)

print(c)

print(c[1].sum())

```

# corrcoef define the relation between the two data
# 0-means no relation ,-1 means inversely proportional, 1 means propostional relantion

```python
tobacco = [30,40,50,10,20,60]
deaths= [100, 200, 300, 25, 50, 400]
print(np.corrcoef(tobacco, deaths))
```
